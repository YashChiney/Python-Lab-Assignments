```python
import pandas as pd
```

```python
#To create a series
num=[1,2,3,4,5,6,7]
series=pd.Series(num)
print(series)
```

```python
#To print specified index(natural)
print(series[0])
```

```python
#To print specified index(given by programmer)
m=['a','b','c','d','e','f','g']
a=[1,2,3,4,5,6,7]
k=pd.Series(a,index=m)
print(k)
print(k['a'])
```

```python
b= [1,2,3]
var = pd.Series(a, index = ["x", "y", "z"])
print(var)
```

```python
#To create a dataframe
d={'Name':['Yash','Parth','Ayush'],'Roll no':[1,2,3],'Course':['CSE AI/DS','BBA','CSE']}
c=pd.DataFrame(d,index=[0,1,2])
print(c)
```

```python
#To print specified rows
print(c.loc[1])
print('\n',c.loc[[0,2]])
```
