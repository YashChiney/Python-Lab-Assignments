```python
import re
```

```python
#To find if numbers from [0-9] are in the text
txt="I usually wake up at 6 am , but today I woke up at 4 am"
x=re.findall("[0-9]",txt)
print(x)
```

```python
#To find the number of lower case characters in text
txt="Today is a very sunny day , I think we should go to Aryabhatta building"
x=re.findall("[a-z]",txt)
print(x)
print("\n Number of lower case characters in the text are:",len(x))
```

```python
#To find the number of upper case characters in text
txt="Today is a very sunny day , I think we should go to Aryabhatta building"
x=re.findall("[A-Z]",txt)
print(x)
print("\n Number of lower case characters in the text are:",len(x))
```
