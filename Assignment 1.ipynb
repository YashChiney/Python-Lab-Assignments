```python
#List
list1=[1,2,3,4,5]
#To add an item
list1.append(6)
print(list1)
#To remove an item
list1.pop(2)
print(list1)
#To update an item
list1[1]=8
print(list1)
```

```python
#Tuple
tuple1=(1,2,3,4,5)
#To add,remove , update items
list1=list(tuple1)
list1.append(6)#add
list1.remove(2)#remove
list1[1]=13#update
tuple2=tuple(list1)
print(tuple2)
```
