```python
#To make a dictionary
dict1={'name':'Yash','Roll':'1'}
print(dict1)
```

```python
#To know number of items
print(len(dict1))
```

```python
#To print data type
print(type(dict1))
```

```python
#To construct a dictionary
dict2=dict(name="Parth",age=18, country='France')
print(dict2)
```

```python
#To print specific key
x=dict1['name']
print(x)
y=dict1.get('Roll')
print(y)
```

```python
#To print name of keys
z=dict1.keys()
print(z)
```

```python
#To print values
m=dict1.values()
print(m)
```

```python
#To print both key and values
k=dict1.items()
print(k)
```

```python
#To update items
dict1.update({'Roll':2})
print(dict1)
dict1['name']='NONE'
print(dict1)
```

```python
#To remove an item
dict1.pop('Roll')
print(dict1)
```
