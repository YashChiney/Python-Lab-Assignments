```python
#9
l = []
for i in range(20):
    x = int(input("Enter number: "))
    l.append(x)
print(l)
even = 0
odd = 0
positive = 0
negative = 0
for i in range(len(l)):
    if l.count(l[i]) > 1:
        print(l[i], "is repeated at index", i)
    if l[i] % 2 == 0:
        even += 1
    else:
        odd += 1
    if l[i] > 0:
        positive += 1
    elif l[i] < 0:
        negative += 1
print("No. of even elements:", even)
print("No. of odd elements:", odd)
print("No. of positive elements:", positive)
print("No. of negative elements:", negative)
```
