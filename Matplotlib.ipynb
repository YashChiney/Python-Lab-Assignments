```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
#To read csv
df=pd.read_csv("sample.csv")
print(df)
```

```python
#To make line chart
x=df['Name']
y=df["Python"]
plt.plot(x,y,ls=":")
```

```python
#Customisation 1
x=df['Name']
y=df["Python"]
plt.plot(x,y,"o",ls=":",mec='blue',mfc='green')
```

```python
#Customisation 2
x=df['Name']
y=df["Python"]
plt.xlabel("Name")
plt.ylabel("Python marks")
plt.title("Python marks comparison")
plt.plot(x,y,"o:g")
```

```python
#To make scatter plot
x=df['Name']
y=df["Math"]
c=['blue','green','red','hotpink','cyan','orange']
s=[1,20,30,40,50,60]
plt.scatter(x,y,c=c,s=s)
plt.xlabel("Name")
plt.ylabel("Math marks")
plt.title("Math marks comparison")
```

```python
#To make pie chart
x=df['Name']
y=df["Python"]
e=(0.5,0,0,0,0.2,0)
plt.pie(y,labels=x,explode=e,shadow=True)
plt.legend()
plt.title("Python marks comparison")
```

```python
#To make bar chart
x=df['Name']
y=df["Python"]
plt.bar(x,y)
plt.title("Python marks comparison")
```

```python
#Horizontal bar
x=df['Name']
y=df["Python"]
plt.barh(x,y)
plt.title("Python marks comparison")
```

```python
#To make heatmap
import seaborn as sns
correlation = df[["Math", "Python", "Age"]].corr()
print("Correlation Matrix:")
print(correlation)

sns.heatmap(correlation, annot=True, cmap="Blues", vmin=-1, vmax=1)
plt.title("Correlation Heatmap")
plt.show()
```
