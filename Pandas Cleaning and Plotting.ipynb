```python
#Making dataframe and reading data
import pandas as pd
x=pd.read_csv("titanic.csv")
print(x)
df=pd.DataFrame(x)
print(df)
```

```python
#Checking for null values
df.isnull()
```

```python
#Summing up null values
df.isnull().sum()
```

```python
#Removing null values via dropping
y=df.dropna()
y.isnull().sum()
```

```python
#Removing null values via filling
z=df.fillna(12)
z.isnull().sum()
```

```python
#Filling in specific cells
v=df.fillna({'Age':13})
v.isnull().sum()
```

```python
#Filling with mean , median or mode
n=df['Age'].mean() #median() or mode() can also be used
k=df.fillna({'Age':n})
print(k)
```

```python
#To check for duplicated values
print(df.duplicated())
```

```python
#Plotting the dataframe
import matplotlib.pyplot as plt
df.plot()
plt.show()
df.plot(kind='scatter',x='Age',y='Fare')
```
